```python
from faker import Faker
import pandas as pd
import random
import pyodbc
```

```python
fake = Faker()

```

```python
# Members

membership_types = ["Basic", "Standard", "Premium"]
genders = ["Male", "Female"]

members = []

for member_id in range(1, 2001):

    name = fake.name()
    email = fake.email()
    phone = fake.numerify("###-###-####")
    gender = random.choice(genders)
    membership = random.choice(membership_types)

    # Join date
    join_date = fake.date_between(
        start_date="-3y",
        end_date="today"
    )

    # Membership period
    membership_start_date = join_date

    membership_end_date = membership_start_date + pd.Timedelta(
        days=random.choice([30, 90, 180, 365])
    )

    members.append({
        "MemberID": member_id,
        "Name": name,
        "Email": email,
        "Phone": phone,
        "Gender": gender,
        "JoinDate": join_date,
        "MembershipType": membership,
        "MembershipStartDate": membership_start_date,
        "MembershipEndDate": membership_end_date
    })

# Convert to DataFrame
members_df = pd.DataFrame(members)

print("Members created:", len(members_df))

# Display first 10 rows
display(members_df.head(10))
```

```python
# Trainers

specializations = [
    "Strength Training",
    "CrossFit",
    "Yoga",
    "Boxing",
    "Nutrition",
    "Cardio",
    "Personal Training",
    "Zumba",
    "Functional Training"
]

trainers = []

for trainer_id in range(1, 51):

    name = fake.name()
    phone = fake.numerify("###-###-####")
    specialization = random.choice(specializations)

    hire_date = fake.date_between(
        start_date="-5y",
        end_date="today"
    )

    trainers.append({
        "TrainerID": trainer_id,
        "Name": name,
        "Specialization": specialization,
        "Phone": phone,
        "HireDate": hire_date
    })

trainers_df = pd.DataFrame(trainers)

print("Trainers created:", len(trainers_df))

display(trainers_df.head(10))
```

```python
# Classes

class_names = [
    "Yoga",
    "CrossFit",
    "Boxing",
    "Zumba",
    "Strength Training",
    "Cardio",
    "HIIT",
    "Pilates",
    "Spinning",
    "Functional Training",
    "Kickboxing",
    "Body Pump",
    "Aerobics",
    "Mobility",
    "Circuit Training"
]

schedules = [
    "08:00 AM",
    "10:00 AM",
    "12:00 PM",
    "04:00 PM",
    "06:00 PM",
    "07:00 PM",
    "08:00 PM"
]

classes = []

for class_id in range(1, 101):

    class_name = random.choice(class_names)

    # Connect Class with Trainer
    trainer_id = random.randint(1, 50)

    schedule = random.choice(schedules)

    capacity = random.randint(10, 30)

    classes.append({
        "ClassID": class_id,
        "ClassName": class_name,
        "TrainerID": trainer_id,
        "Schedule": schedule,
        "Capacity": capacity
    })

classes_df = pd.DataFrame(classes)

print("Classes created:", len(classes_df))

display(classes_df.head(10))
```

```python
# Class Bookings

booking_statuses = [
    "Booked",
    "Completed",
    "Cancelled",
    "No Show"
]

bookings = []

for booking_id in range(1, 30001):

    # Existing Member
    member_id = random.randint(1, 2000)

    # Existing Class
    class_id = random.randint(1, 100)

    # Get member membership period
    member_data = members_df[
        members_df["MemberID"] == member_id
    ].iloc[0]

    start_date = member_data["MembershipStartDate"]
    end_date = member_data["MembershipEndDate"]

    # Booking inside membership period
    booking_date = fake.date_between(
        start_date=start_date,
        end_date=end_date
    )

    booking_status = random.choices(
        booking_statuses,
        weights=[50, 30, 10, 10]
    )[0]

    bookings.append({
        "BookingID": booking_id,
        "MemberID": member_id,
        "ClassID": class_id,
        "BookingDate": booking_date,
        "BookingStatus": booking_status
    })

bookings_df = pd.DataFrame(bookings)

print("Bookings created:", len(bookings_df))

display(bookings_df.head(10))
```

```python
# Attendance

attendance_statuses = [
    "Present",
    "Absent",
    "Late"
]

attendance = []


selected_bookings = bookings_df.sample(
    n=20000,
    random_state=42
)

for attendance_id, (_, booking) in enumerate(
    selected_bookings.iterrows(),
    start=1
):

    member_id = booking["MemberID"]
    class_id = booking["ClassID"]
    attendance_date = booking["BookingDate"]

    status = random.choices(
        attendance_statuses,
        weights=[70, 20, 10]
    )[0]

    attendance.append({
        "AttendanceID": attendance_id,
        "MemberID": member_id,
        "ClassID": class_id,
        "AttendanceDate": attendance_date,
        "Status": status
    })

attendance_df = pd.DataFrame(attendance)

print("Attendance created:", len(attendance_df))

display(attendance_df.head(10))
```

```python
# Payments

payment_methods = [
    "Cash",
    "Credit Card",
    "Bank Transfer"
]

payment_statuses = [
    "Paid",
    "Failed",
    "Pending"
]

payments = []

for payment_id in range(1, 8001):

    # Existing Member
    member_id = random.randint(1, 2000)

    # Get member data
    member_data = members_df[
        members_df["MemberID"] == member_id
    ].iloc[0]

    start_date = member_data["MembershipStartDate"]
    end_date = member_data["MembershipEndDate"]

    # Payment date inside membership period
    payment_date = fake.date_between(
        start_date=start_date,
        end_date=end_date
    )

    membership = member_data["MembershipType"]

    # Payment amount based on membership type
    if membership == "Basic":
        amount = random.choice([500, 550, 600])

    elif membership == "Standard":
        amount = random.choice([800, 850, 900])

    else:
        amount = random.choice([1200, 1250, 1300])

    payment_method = random.choice(payment_methods)
    payment_status = random.choices(
        payment_statuses,
        weights=[85, 10, 5]
    )[0]

    payments.append({
        "PaymentID": payment_id,
        "MemberID": member_id,
        "PaymentDate": payment_date,
        "Amount": amount,
        "PaymentMethod": payment_method,
        "PaymentStatus": payment_status
    })

payments_df = pd.DataFrame(payments)

print("Payments created:", len(payments_df))

display(payments_df.head(10))
```

```python
members_df.to_csv("members.csv", index=False)
trainers_df.to_csv("trainers.csv", index=False)
classes_df.to_csv("classes.csv", index=False)
bookings_df.to_csv("class_bookings.csv", index=False)
attendance_df.to_csv("attendance.csv", index=False)
payments_df.to_csv("payments.csv", index=False)

print("All files saved successfully!")
```
